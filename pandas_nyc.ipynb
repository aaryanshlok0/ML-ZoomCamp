# CLEANING DATA


```python
df.isnull().sum()
```

```python
df2=df.dropna() #dropping all the rows having null values , now only 71 rows are left rest 29 had null values in a column or another so they got removed
```

```python
df2.isnull().sum() #the problem is that 29 rows got deleted out of 100 so not the  rigth way
```

```python
df.dropna(subset=['city']) #10 rows got dropped with null values in city
```

```python
#filling those values instead of dropping entire row which can cause issues and deletion of imp data
```

```python
df['age']=df['age'].fillna(df['age'].median()) #df['age'] tells python to get the age series and then the rest makes calc median of non missing values and fill it
```

```python
df.isnull().sum() #inplace=true makes it into real data now null age = 0
```

```python
df['salary']=df['salary'].str.replace(",","",regex=False)
df['salary']=pd.to_numeric(df['salary'],errors='coerce')
df['salary']=df['salary'].fillna(df['salary'].median())
```

```python
df['salary'].dtype #can also use df.dtypes
```

```python
df.isnull().sum() #0 null values in salary and age column
```

```python
#now to fill dept its not a numeric value so we would see first what kinda values are in there
```

```python
df['department'].value_counts()
#its a categorical col not a continuous(values ) , we fill categorical col with mode
```

```python
df['department']=df['department'].fillna(df['department'].mode()[0]) #this [0] is as there are both hr and marketing with same count so we gotta select one out of these to fill
```

```python
df['city']=df['city'].fillna(df['city'].mode()[0])
```
